```python

import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("/kaggle/input/datasets/meghanakandhi/pcos-dataset/PCOS_synthetic_50000_44cols.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Target
target = "PCOS (Y/N)"

# Remove identifiers
df = df.drop(
    columns=["Sl. No", "Patient File No."],
    errors="ignore"
)

# Check dataset
print("Dataset shape:", df.shape)
print(df.head())
```

```python

print("Cycle values:")
print(df["Cycle(R/I)"].value_counts(dropna=False))

print("\nTarget values:")
print(df["PCOS (Y/N)"].value_counts(dropna=False))

print("\nData types:")
print(df.dtypes)
```

```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# STEP 5: DATA PREPROCESSING

early_features = [
    "Age (yrs)",
    "Weight (Kg)",
    "Height(Cm)",
    "BMI",
    "Cycle(R/I)",
    "Cycle length(days)",
    "Weight gain(Y/N)",
    "hair growth(Y/N)",
    "Skin darkening (Y/N)",
    "Hair loss(Y/N)",
    "Pimples(Y/N)",
    "Fast food (Y/N)",
    "Reg.Exercise(Y/N)"
]

clinical_features = [
    "Pulse rate(bpm)",
    "RR (breaths/min)",
    "Hb(g/dl)",
    "Pregnant(Y/N)",
    "No. of abortions",
    "FSH(mIU/mL)",
    "LH(mIU/mL)",
    "FSH/LH",
    "Hip(inch)",
    "Waist(inch)",
    "Waist:Hip Ratio",
    "TSH (mIU/L)",
    "AMH(ng/mL)",
    "PRL(ng/mL)",
    "Vit D3 (ng/mL)",
    "PRG(ng/mL)",
    "RBS(mg/dl)",
    "BP _Systolic (mmHg)",
    "BP _Diastolic (mmHg)",
    "Follicle No. (L)",
    "Follicle No. (R)",
    "Avg. F size (L) (mm)",
    "Avg. F size (R) (mm)",
    "Endometrium (mm)"
]

early_features = [
    c for c in early_features if c in df.columns
]

clinical_features = [
    c for c in clinical_features if c in df.columns
]

extended_features = early_features + clinical_features

# Input and target
X_early = df[early_features].copy()
X_extended = df[extended_features].copy()
y = df[target].copy()

# Split before fitting preprocessing
X_train_early, X_test_early, y_train, y_test = train_test_split(
    X_early,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train_ext, X_test_ext, _, _ = train_test_split(
    X_extended,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Identify numerical and categorical columns
numeric_early = X_train_early.select_dtypes(
    include=np.number
).columns.tolist()

categorical_early = X_train_early.select_dtypes(
    exclude=np.number
).columns.tolist()

numeric_ext = X_train_ext.select_dtypes(
    include=np.number
).columns.tolist()

categorical_ext = X_train_ext.select_dtypes(
    exclude=np.number
).columns.tolist()

# Preprocessing for early model
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_early = ColumnTransformer([
    ("num", numeric_transformer, numeric_early),
    ("cat", categorical_transformer, categorical_early)
])

preprocessor_extended = ColumnTransformer([
    ("num", numeric_transformer, numeric_ext),
    ("cat", categorical_transformer, categorical_ext)
])

# Fit only on training data
X_train_early_processed = preprocessor_early.fit_transform(
    X_train_early
)

X_test_early_processed = preprocessor_early.transform(
    X_test_early
)

X_train_ext_processed = preprocessor_extended.fit_transform(
    X_train_ext
)

X_test_ext_processed = preprocessor_extended.transform(
    X_test_ext
)

print("Early training:", X_train_early_processed.shape)
print("Early testing:", X_test_early_processed.shape)

print("Extended training:", X_train_ext_processed.shape)
print("Extended testing:", X_test_ext_processed.shape)
```

```python
print("Missing values after preprocessing:")

print(
    np.isnan(X_train_early_processed.toarray()
             if hasattr(X_train_early_processed, "toarray")
             else X_train_early_processed).sum()
)

print("\nTarget distribution in training:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTarget distribution in testing:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("All libraries imported successfully!")
```

```python
# Model 1: Logistic Regression
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(
    X_train_early_processed,
    y_train
)

print("Logistic Regression training completed!")
```

```python
# Model 2: Random Forest
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(
    X_train_early_processed,
    y_train
)

print("Random Forest training completed!")
```

```python
# Model 3: XGBoost
xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train_early_processed,
    y_train
)

print("XGBoost training completed!")
```

```python
models = {
    "Logistic Regression": lr_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model
}

results = []

for name, model in models.items():

    y_pred = model.predict(X_test_early_processed)

    y_prob = model.predict_proba(
        X_test_early_processed
    )[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

print(results_df.round(4))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix

# Predict using XGBoost
y_pred_xgb = xgb_model.predict(X_test_early_processed)

# Create confusion matrix
cm = confusion_matrix(y_test, y_pred_xgb)

print("Confusion Matrix:")
print(cm)

# Visualize
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No PCOS", "PCOS"],
    yticklabels=["No PCOS", "PCOS"]
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("XGBoost Confusion Matrix")

plt.show()
```

```python
import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("SHAP version:", shap.__version__)
```

```python
X_shap = X_test_early_processed[:500]

print("SHAP data shape:", X_shap.shape)
```

```python
explainer = shap.TreeExplainer(xgb_model)

shap_values = explainer.shap_values(X_shap)

print("SHAP values calculated successfully!")
print("SHAP values shape:", shap_values.shape)
```

```python
shap.summary_plot(
    shap_values,
    X_shap,
    feature_names=early_features,
    show=True
)
```

```python
# Explain the first test record
shap_explanation = explainer(X_shap[:1])

# Display waterfall plot
shap.plots.waterfall(
    shap_explanation[0],
    max_display=13
)
```

```python
# Create an explanation with actual feature names
shap_exp = shap.Explanation(
    values=shap_values[0],
    base_values=explainer.expected_value,
    data=X_shap[0],
    feature_names=early_features
)

# Generate waterfall plot
shap.plots.waterfall(
    shap_exp,
    max_display=13
)
```
